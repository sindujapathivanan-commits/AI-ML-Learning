```python
import numpy as np
import pandas as pd

#1.Cuatomer Data Set
print("==========Cuatomer Data Set=======")
file_path = r"d:\Sinduja\Appton\Pandas Data set Ass 21\Customer Dataset.csv"  # update extension if your file is csv

df = pd.read_csv(file_path)
#print(df)

print("==========Part 1,2=======")
#1.Shape
print("==========1.Shape=======")
print(df.shape)

#2.Columns
print("==========2.Columns=======")
print(df.columns)


#3.Data types
print("==========3.Data types=======")
print(df.dtypes)


#4.Missing values
print("==========4.Missing values=======")
print(df.isnull)


#5.Duplicate count
print("==========5.Duplicate count=======")
print(df.duplicated)


#6.Statistical summary
print("==========6.Statistical summary=======")
print(df.describe())
print("Count:", df['Total Charges'].count())
print("Min:", df['Total Charges'].min())
print("Max:", df['Total Charges'].max())


print("==========Part 3 =======")

#1.Missing values
print("==========1.Missing values=======")
print(df.isnull().sum())

print("==========fill the Missing values=======")
# Numeric columns -> fill with mean
df.fillna(df.mean(numeric_only=True), inplace=True)

# Text columns -> fill with mode (most frequent value)
for col in df.select_dtypes(include='object'):
    df[col].fillna(df[col].mode()[0], inplace=True)
    
# Replace blanks and spaces with NaN
df = df.replace(r'^\s*$', np.nan, regex=True)

# Fill numeric columns with mean
for col in df.select_dtypes(include='number'):
    df[col] = df[col].fillna(df[col].mean())

# Fill text columns with mode
for col in df.select_dtypes(include='object'):
    df[col] = df[col].fillna(df[col].mode()[0])

print(df.isnull().sum())

#2.Duplicate records
print("==========2.Duplicate records=======")
print(df.duplicated().sum())

df.drop_duplicates(keep='first') #Keep the first matching row.


#3.Invalid ages
print("==========3.Invalid ages=======")
invalid_records = df[df["Age"] < 0]
print(invalid_records)

df["Age"] = df["Age"].fillna(df["Age"].median())  # Fill the ages with correct values


#4.Invalid  Total Charges
print("==========4.Invalid Total Charges=======")
# Convert to numeric first
df["Total Charges"] = pd.to_numeric(df["Total Charges"], errors="coerce")

# Replace negative values with 0
df.loc[df["Total Charges"] < 0, "Total Charges"] = 0

# Check again
print((df["Total Charges"] < 0).sum())


#5.Inconsistent categorical values
print("==========5.Inconsistent categorical values=======")

print("Gender:", df["Gender"].unique())
print("Contract Type:", df["Contract Type"].unique())
print("Payment Method:", df["Payment Method"].unique())
print("Churn:", df["Churn"].unique())

# Before cleaning
for col in df.select_dtypes(include="object"):
    print(f"{col}:", df[col].unique())

# Clean categorical columns
for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip().str.title()

# After cleaning
for col in df.select_dtypes(include="object"):
    print(f"\n{col}:")
    print(df[col].unique())
    
#After-Cleaning Report
print("==========After-Cleaning Report=======")

# Missing values after cleaning
missing_report = df.isnull().sum()

# Duplicate records remaining
duplicate_count = df.duplicated().sum()

# Dataset shape
rows, cols = df.shape

# Data types
dtypes = df.dtypes

# Basic statistical summary
stats = df.describe(include='all')

print("Missing Values:\n", missing_report)
print("\nDuplicates:", duplicate_count)
print("\nShape:", (rows, cols))
print("\nData Types:\n", dtypes)
print("\nStatistics:\n", stats)

# Prepare ML Dataset Separate
print("==========Prepare ML Dataset Separate=======")

import pandas as pd

# 1. Load dataset
df = pd.read_csv(file_path)
print("\nOriginal Dataset Shape:", df.shape)

# 2. Clean dataset
before_cleaning = len(df)

df = df.drop_duplicates()
after_duplicates = len(df)

df = df.dropna()
after_cleaning = len(df)

print("\n========== Cleaning Summary ==========")
print("Original Customers:", before_cleaning)
print("Duplicates Removed:", before_cleaning - after_duplicates)
print("Rows Removed Due to Missing Values:", after_duplicates - after_cleaning)
print("Clean Customers:", len(df))

# 3. Save cleaned file
df.to_csv("clean_customer_data.csv", index=False)
print("\nCleaned file saved: clean_customer_data.csv")

# 4. Separate features and target
X = df.drop("Churn", axis=1)
y = df["Churn"]

print("\n========== ML Dataset ==========")
print("Features (X) Shape:", X.shape)
print("Target (y) Shape:", y.shape)
print("\nTarget Column: Churn")
print("Churn Values:")
print(y.value_counts())

# 5. Save ML-ready data
df.to_csv("ml_ready_customer_data.csv", index=False)
print("\nML-ready file saved: ml_ready_customer_data.csv")
      
      
      
# Final Report
print("========== — Final Report=======")

# 1. Load original dataset
import pandas as pd

original_df = pd.read_csv(file_path)

# 2. Total customers before cleaning
total_customers = len(original_df)

# 3. Missing values before cleaning
missing_values_before = original_df.isnull().sum().sum()

# 4. Count duplicates before removing them
removed_duplicates = original_df.duplicated().sum()

# 5. Clean the dataset
df = original_df.drop_duplicates().dropna().copy()

# 6. Clean customers
clean_customers = len(df)

# 7. Missing values after cleaning
missing_values_after = df.isnull().sum().sum()

# 8. Churned customers
churned_customers = (
    df["Churn"].astype(str).str.strip().str.title() == "Yes"
).sum()

# 9. Churn rate
churn_rate = (churned_customers / clean_customers * 100
              if clean_customers > 0 else 0)

# 10. Average monthly charges
average_monthly_charges = pd.to_numeric(
    df["Monthly Charges"].astype(str).str.replace(",", ""),
    errors="coerce"
).mean()

# 11. Highest churn segment
churn_data = df.copy()
churn_data["Churn"] = (
    churn_data["Churn"].astype(str).str.strip().str.title()
)

churn_by_segment = (
    churn_data.groupby("Contract Type")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
)

highest_churn_segment = (
    churn_by_segment.idxmax()
    if not churn_by_segment.empty
    else "Not Available"
)

# 12. Display final report
print("Total Customers:", total_customers)
print("Clean Customers:", clean_customers)
print("Removed Duplicates:", removed_duplicates)
print("Missing Values Before:", missing_values_before)
print("Missing Values After:", missing_values_after)
print("Churned Customers:", churned_customers)
print(f"Churn Rate: {churn_rate:.2f}%")

print(
    "Average Monthly Charges:",
    round(average_monthly_charges, 2)
    if pd.notna(average_monthly_charges)
    else "Not Available"
)

print("Highest Churn Segment:", highest_churn_segment)

print("============================================")
```
