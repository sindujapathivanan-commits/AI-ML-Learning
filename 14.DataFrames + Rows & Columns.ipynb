```python
import numpy as np
import pandas as pd

#All students
print("\n=====1.All students====")
data = {
'Employee ID' :[100, 200, 300, 400, 500, 600],
'Name': ['Mala','Vaani','Leela','Varun','Pathi','Nilu'],
'Age': [16, 15, 19, 22, 23, 18],
'Department': ['IT', 'HR', 'Admin', 'Sales', 'IT', 'Admin'],
'Salary': [34456, 56043, 87045, 78234, 89056, 89003],
'Experience': [7, 5, 9, 7, 5, 8],
'Performance Score': [3, 5, 4, 6, 5, 6]
}

df = pd.DataFrame(data)
print(df)

#Display selected columns.
print("\n=====2.Display selected columns====")
df.Salary

#Display first 5 employes.
print("\n=====3.Display first 5 employees====")
print(df.head(5))

#Display Last 5 employes.
print("\n=====4.Display last 5 employees====")
print(df.tail(5))

#Display Add a Bonus column.
print("\n=====4.Add a Bonus column====")
df['Bonus'] = 5000
print(df)

#Rename Salary to Base_Salary
print("\n=====5.Rename Salary to Base_Salary====")
df = df.rename(columns={'Salary':'Base_Salary'})
print(df)

#Rename Remove one unnecessary column.
print("\n=====6.Remove one unnecessary column.====")
df.drop('Age', axis=1)
print(df)

#Select one employee using loc.
print("\n=====7.Select one employee using loc====")
print(df.loc[2])   #selects the entire 2nd row

#Select one employee using iloc
print("\n====8.Select one employee using il0c===")
print(df.iloc[2])    #position 2                         #You know the position.

#Select one employee
print("\n====Table===")
print(df.loc[2, "Name"])   # Leela
print(df.iloc[2, 1])     

#Select Total_Salary column
print("\n====8.Total_Salary column===")
print(df['Base_Salary'].sum())




```
