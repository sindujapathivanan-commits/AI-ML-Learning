```python
import numpy as np

sales = np.array([
    
 # Each row is: (Product ID, Monthly Sales, Revenue, Profit).
 
(1, 412, 14817, 4826),
(2, 202, 4270, 1329),
(3, 299, 11078, 2154),
(4, 181, 6087, 749),
(5, 490, 5370, 1414),
(6, 306, 9900, 2370),
(7, 327, 6661, 1268),
(8, 197, 5240, 818),
(9, 252, 5377, 1084),
(10, 413, 15580, 3384),
(11, 219, 6048, 1011),
(12, 188, 3059, 529),
(13, 455, 14462, 3100),
(14, 169, 6964, 2194),
(15, 472, 10504, 2589)
        
])

print(sales)


#Average Revenue
avg_revenue = sales[:, 2].mean()
print("Average Revenue:", avg_revenue)

#Highest Sales
highest_sales = sales[sales[:, 1].argmax()]
print("Highest Sales -> Product:", highest_sales)

#Lowest Sales
lowest_sales = sales[sales[:, 1].argmin()]
print("Lowest Sales -> Product:", lowest_sales)

#Profit Percentage (Profit / Revenue * 100) for each product
profit_percentage = (sales[:, 3] / sales[:, 2]) * 100
print("Profit Percentage per product:", np.round(profit_percentage, 2))

#Best Selling Product (highest Monthly Sales)
best_selling = sales[sales[:, 1].argmax()]
print("Best Selling Product:", best_selling)

#Worst Selling Product (lowest Monthly Sales)
worst_selling = sales[sales[:, 1].argmin()]
print("Worst Selling Product:", worst_selling)

#Bonus
#Display the Top 5 products based on revenue.
top5_indices = sales[:, 2].argsort()[::-1][:5]
top5_products = sales[top5_indices]
print("Top 5 Products by Revenue:\n", top5_products)
```
