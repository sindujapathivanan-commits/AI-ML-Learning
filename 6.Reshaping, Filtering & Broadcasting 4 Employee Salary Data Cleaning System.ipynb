```python
import numpy as np

#Prepare employee salary data

# Columns: Employee_ID, Age, Department, Salary, Experience, Performance_Score
#dept_names = {1: "IT", 2: "Finance", 3: "HR", 4: "Marketing"}

employees = np.array([
    [101, 25, 1, 50000,  2, 85],
    [102, 32, 2, 60000,  5, 92],
    [103, 45, 3, -70000, 10, 95],
    [104, 29, 4, 55000,  4, 88],
    [105, 61, 1, 80000, 15, 97],
])

print("\n===== 1. ORIGINAL EMPLOYEE DATA =====")
print(employees)



#Increase salary by 10%.

salary = np.array([50000, 60000, -70000, 55000, 80000])

salary = salary * 1.10

print("\n===== 2. SALARY AFTER 10% INCREASE =====")

print(salary)

#Replace negative salaries.

salary = np.array([50000, 60000, -70000, 55000, 80000])

salary[salary == -70000] = 80000

print("\n===== 3. ORIGINAL SALARY =====")
print(salary)


#Filter employees with performance > 90.

performance = np.array([85, 92, 95, 88, 97])

result = performance[performance > 90]

print("\n===== 4. Performance > 90 =====")
print(result)

#Find employees earning above average salary.

salary1 = np.array([50000, 60000, 80000, 55000, 80000])

total = ([50000 + 60000 + 80000 + 55000 + 80000])

print(total) #total salary

average = np.mean(salary1) # average

print("\n===== 5. Average salary =====")

print(average)

#Reshape dataset.
employees1 = np.array([
    [101, 25, 1, 50000,  2, 85],
    [102, 32, 2, 60000,  5, 92],
    [103, 45, 3, 80000, 10, 95],
    [104, 29, 4, 55000,  4, 88],
    [105, 61, 1, 80000, 15, 97],
])

print(employees1.shape) # shape

result2 = employees1.reshape(6, 5)

print("\n===== 6. Reshape dataset =====")

print(result2)

#Transpose dataset.

result3 = employees1.T

print("\n===== 7. Transpose dataset =====")

print(result3)


```
