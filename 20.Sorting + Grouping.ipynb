```python
import pandas as pd
import numpy as np

#1.Sales Performance Analyzer
print("==========Sales Performance Analyzer=======")
file_path = r"d:\Sinduja\Appton\Sales Performance Analyzer\Sales Performance Analyzer.csv"  # update extension if your file is csv

df = pd.read_csv(file_path)
print(df)

#Revenue by region.
print("==========1.Revenue by region=======")
print(df.groupby("Region")["Revenue"].sum())




#Revenue by category.
print("==========2.Revenue by category=======")
print(df.groupby("Category")["Revenue"].sum())




#Profit by salesperson.
print("==========3.Profit by salesperson=======")
print(df.groupby("Salesperson")["Profit"].sum())




#Top salesperson.
print("==========4.Top salesperson.=======")
print(df.loc[df["Profit"].idxmax(), ["Salesperson", "Profit"]])




#Best-performing region.
print("==========5.Best-performing region.======")
print(df.loc[df["Profit"].idxmax(), ["Region", "Profit"]])




#Best-selling category.
print("==========6.Best-selling category.======")
print(df.loc[df["Profit"].idxmax(), ["Category", "Profit"]])





#Rank regions by total profit
print("==========7.Rank regions by total profit=======")
print(df.groupby("Region")["Profit"].sum().sort_values(ascending=False))




```
