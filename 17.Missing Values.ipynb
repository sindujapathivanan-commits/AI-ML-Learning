```python
import numpy as np
import pandas as pd

#1.Patient Details
print("\n=====Displays dataset====")
file_path = r"d:\Sinduja\Appton\Python\patient_records_500.xlsx"  # update extension if your file is .xlsx

df = pd.read_excel(file_path)
print(df)

#2.Detect missing values
print("\n===== Detect missing values====")
df.isnull().sum().sum()

#3.Display missing count per column
print("\n=====Display missing count per column====")
df.isnull().sum()

#4.Calculate missing percentage
print("\n=====Calculate missing percentage.====")
missing_percentage = df.isnull().sum()/len(df)*100              #Missing count ÷ Total rows × 100
print(missing_percentage.round(0).astype(int))               #astype() = Change data type (Float → Integer)

#5.Fill numerical missing values appropriately/Number missing
print("\n=====Fill numerical missing values appropriately====")

#Find mean of all numerical columns
print("\n=====Find mean of all numerical columns===")
Mean = df.mean(numeric_only=True)
print(Mean.round(0).astype(int))    

#Fill missing values in every numerical column
print("\n=====Fill missing values in every numerical column===")
df=df.fillna(df.mean(numeric_only=True))
print(df)     

#6. Fill categorical missing values using an appropriate strategy./Text Missing
print("\n===== Fill categorical missing values using an appropriate strategy.====")
   
#Find moad for gender columns
print("\n=====Find moad for gender columns===")
print(df["Gender"].mode())

#Find moad for Patient ID columns                                   #Find moad for Patient ID columns
print("\n=====Find moad for Patient ID columns===")
print(df["Patient ID"].mode())

print("\n=====fill missing Patient ID  columns===")
df["Patient ID"]=df["Patient ID"].fillna(df["Patient ID"].mode()[0])
print(df)


#fill missing values gender columns
print("\n=====fill missing values gender columns===")
df["Gender"]=df["Gender"].fillna(df["Gender"].mode()[0])
print(df)

#7.Missing values compare
print("\n=====Missing values compare===")

#Original data copy
df_before = pd.read_excel(file_path)
print(df_before)

print("Before Cleaning:")
print(df_before.isnull().sum())

print("After Cleaning:")
print(df.isnull().sum())

#Total Records
print("\n=====Total Records:===")
print(len(df))


#Cleaning Status:
print("\n=====Cleaning Status:===")
print("Total Records:", len(df))
print("Total Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())

```
