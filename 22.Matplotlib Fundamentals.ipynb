```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


students = [
    "Aisha Perera", "Nimal Silva", "Kavindi Fernando", "Ruwan Jayasinghe",
    "Tharushi Wickramasinghe", "Dilshan Rajapaksa", "Sanduni Gunawardena",
    "Hasitha Bandara", "Nethmi Senanayake", "Chamara Dissanayake",
    "Ishara Kumari", "Lahiru Madushanka", "Mihiri Samarasinghe",
    "Pasindu Herath", "Sachini Abeysekara", "Tharindu Weerasinghe",
    "Oshadi Ranasinghe", "Kasun Liyanage", "Dinithi Amarasinghe",
    "Buddhika Rathnayake",
]
 
math    = [78, 65, 92, 55, 81, 70, 88, 62, 95, 74, 83, 59, 90, 68, 76, 85, 67, 73, 86, 52]
science = [82, 70, 88, 60, 79, 74, 91, 58, 93, 69, 85, 64, 87, 72, 80, 78, 71, 66, 89, 57]
python  = [85, 60, 95, 48, 87, 68, 90, 66, 97, 72, 79, 55, 92, 75, 83, 88, 63, 78, 84, 50]
ai      = [80, 58, 93, 50, 84, 72, 89, 61, 96, 70, 81, 57, 91, 77, 79, 86, 65, 74, 87, 53]

plt.figure(figsize=(12, 6))
 
plt.plot(students, math, marker="o", label="Math")
plt.plot(students, science, marker="o", label="Science")
plt.plot(students, python, marker="o", label="Python")
plt.plot(students, ai, marker="o", label="AI")
 
plt.title("Student Marks by Subject")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.xticks(rotation=60, ha="right") #என்பது X-axis labels-ஐ சாய்வாகவும் வலது பக்கம் align செய்தும் காட்டும். rotation = சாய்வு; ha = horizontal alignment., student names நீளமாக இருப்பதால், அவை ஒன்றோடு ஒன்று overlap ஆகாமல் படிக்க உதவும்.
plt.legend() #Chart-ல் எந்த line எந்த subject-ஐக் குறிக்கிறது என்பதைக் காட்டும்.
plt.tight_layout() #Chart spacing-ஐச் சரிசெய்யும்
 
plt.show()


# ---------------- Pie Chart: Grade Distribution ----------------
# Pie Chart: Grade Distribution

pie_labels = ["A", "B", "C", "D", "F"]
pie_values = [5, 6, 4, 3, 2]

plt.figure(figsize=(7, 7)) # Easy way to remember: figsize = (width, height).

plt.pie(
    pie_values,
    labels=pie_labels,
    autopct="%1.1f%%", #display the percentage value on each slice of the pie chart.
    startangle=90
)

plt.title("Grade Distribution")
plt.show()


# Bar Chart: Student vs Average Marks

average_marks = np.mean([math, science, python, ai], axis=0) #student-க்கும் 4 subjects-ன் average marks

plt.figure(figsize=(12, 6))
plt.bar(students, average_marks)

plt.title("Student vs Average Marks")
plt.xlabel("Students")
plt.ylabel("Average Marks")

plt.xticks(rotation=60, ha="right")
plt.tight_layout()
plt.show()
 
```
