```python
import pandas as pd
import numpy as np


#1. Customer Segmentation System
print("==========Customer Segmentation System=======")
file_path = r"d:\Sinduja\Appton\Customer Segment Data Set\Customers_Segmented Data.csv"  # update extension if your file is csv

df = pd.read_csv(file_path)
#print(df)

print(df.columns.tolist())

#1.Customers above age 40.
print("==========1.Customers above age 40=======")
print(df[df["Age"]>40])

#2.Customers with income > 100,000.                                      Dout for with means add =>
print("==========2.Customers with income >= 100,000=======")

df.columns = df.columns.str.strip()
df["Income"] = pd.to_numeric(df["Income"].astype(str).str.replace(",", "", regex=False), errors="coerce")
print(df[df["Income"] >= 100000])

            
#3.Customers spending > 50,000.
print("==========3.3.Customers spending > 50,000.=======")
print(df[df["Income"]>50000])

#4.Colombo customers.
print("==========4.Colombo customers=======")
print(df[df["City"]=="Colombo"])


#5.High-income + high-spending customers.                                   Dout high-spending not available
print("==========5.High-income + high-spending customers=======")
print(df["Income"].min())
print(df["Income"].max())


#6.Customers who have never purchased
print("==========6.Customers who have never purchased=======")
print(df[df["Income"]==0])


#7.Create customer categories
print(df.columns)

print("==========6.customer categories=======")
df["Category"] = df["Income"].apply(
    lambda x: "Premium" if x >= 150000
    else "Regular" if x >= 100000
    else "Basic"
)

print(df[["Income","Gender","Category"]])

```
