```python
import pandas as pd
import numpy as np


#1.Ecommerce Data Set Details
print("\n=====Displays dataset====")
file_path = r"d:\Sinduja\Appton\E-Commerce Data Set\ECommerce Data Set.csv"  # update extension if your file is .xlsx

df = pd.read_csv(file_path)
print(df)

#2.Missing prices
print("\n=====Missing values====")
print(df.isnull().sum().sum())

#3.Display missing count per column
print("\n=====Display missing count per column====")
print(df.isnull().sum())

#Find mean of all numerical columns
print("\n=====Find mean of all numerical columns===")
Mean = df.mean(numeric_only=True)
print(Mean.round(0).astype(int))   

#Fill missing values in every numerical column
print("\n=====Fill missing values in every numerical column===")
df=df.fillna(df.mean(numeric_only=True))
print(df)     


#Find moad for Quantity columns
print("\n=====Find moad for gender columns===")
print(df["Quantity"].mode())

#Find moad for Quantity columns                                   #Find moad for Quantity columns
print("\n=====Find moad for Quantity columns===")
print(df["Quantity"].mode())

print("\n=====fill missing Quantity ID  columns===")
df["Quantity"]=df["Quantity"].fillna(df["Quantity"].mode()[0])
print(df)


#Find moad for Order Date columns
print("\n=====Find moad Order Date columns===")
print(df["Order Date"].mode())

#Find moad for Order Date columns                                   #Find moad for Order Date columns
print("\n=====Find moad for Order Date columns===")
print(df["Order Date"].mode())

print("\n=====fill missing Order Date  columns===")
df["Order Date"]=df["Order Date"].fillna(df["Order Date"].mode()[0])
print(df)





#3.Duplicate orders
print("\n=====Duplicate orders====")
print(df.duplicated().sum())

#Specific rows duplicates check
print("\n===== Duplicate Order IDs =====")
print(df.duplicated(subset="Product").sum())

#4.Invalid quantities
print("\n=====Invalid quantities =====")
print(df[df['Price'] < 0])


#5.Inconsistent city names
print("\n=====Inconsistent city names =====")
print("\n=====See all city names=====")
print(df.City)

print("\n=====See all city Unique names=====")
print(df['City'].unique())

print("\n=====fix differences in uppercase and lowercase=====")
df['City'] = df['City'].str.strip().str.title()

#6.Incorrect date formats
print("\n=====Check data formats=====")
print(df.dtypes)


#Cleaning Status:
print("\n=====Cleaning Status:===")
print("Quantity:", len(df))
print("Price:", df.isnull().sum().sum())
print("City:", df.duplicated().sum())



#7.Missing values compare
print("\n=====Missing values compare===")

#Original data copy
df_before = pd.read_csv(file_path)
print(df_before)

print("Before Cleaning:")
print(df_before.isnull().sum())

print("After Cleaning:")
print(df.isnull().sum())

#Total Records
print("\n=====Total Records:===")
print(len(df))



#8.clean_orders.csv
print("\n=====clean_orders.csv===")
df.to_csv("clean_orders.csv", index=False)


#9. Cleaning Log
print("\n=====Cleaning Log===")
cleaning_log = pd.DataFrame({
    "Transformation": [
        "Missing values checked",
        "Quantity missing values filled",
        "Price missing values filled",
        "Order Date missing values handled",
        "Duplicate rows checked",
        "City names standardized",
        "Order Date format standardized"
    ],
    "Status": [
        "Completed",
        "Completed",
        "Completed",
        "Completed",
        "Completed",
        "Completed",
        "Completed"
    ]
})

print(cleaning_log)

```
