```python
import numpy as np
import pandas as pd

#Patiend Details
print("\n====Patiend Details======")
patiends = {
    'Patiend ID' : [100, 200, 300, 400, 500, 600, 700, 800, 900],
    'Age' : [23, 45, 56, 78, 78, 67, 89, 34, 56],
    'Gender' : ['Male', 'Female', 'Male', 'Male', 'Female', 'Male', 'Male', 'Female', 'Male'],
    'Blood Pressure' : [100, 150, 110, 200, 120, 400, 250, 100, 170],
    'Glucose' : [110, 110, 85, 120, 100, 130, 105, 88, 66],
    'Weight' : [55, 65, 70, 80, 60, 75, 68, -85, 72],
    'Appointment date' : [
        '2026-10-01',
        '2026-10-02',
        '2026-10-03',
        '2026-10-04',
        '2026-10-05',
        '2026-10-06',
        '2026-10-07',
        '2026-10-08',
        '2026-10-09'
        ]    
}

df = pd.DataFrame(patiends)
print(df)

#Dataset shape
print("\n=====Dataset shape====")
print(df.shape)

#Dataset type
print("\n=====Dataset type====")
print(df.dtypes)

#Dataset missing value
print("\n=====Data set missing value====")
df.isnull()
df.isnull().sum()
df[df.isnull().any(axis=1)] #Show me the complete rows that contain missing values.

#Dataset Duplicate
print("\n=====Data set Duplicate value====")
df.drop_duplicates(keep='first') #Keep the first matching row.
df.drop_duplicates(keep='last') #Keep the last matching row.

#Dataset  Invalid records
print("\n=====Data set  Invalid records====")
invalid_records = df[df["Weight"] < 0]
print(invalid_records)

#Dataset  Data Quality Report
print("\n=====Data Quality Report====")
report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Data Type": df.dtypes.astype(str).values, #astype(str) converts the data type names into text.
    "Duplicate Rows": [df.duplicated().sum()] * len(df.columns) #len(df.columns) counts how many columns your dataset has.
})

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
print("\n", report.to_string(index=False)) #to_string() converts the table into readable text. #index=False hides the row numbers on the left
```
